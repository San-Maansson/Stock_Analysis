```python
import numpy as np
import pandas as pd

#Covariance matrix of assets
Sigma = np.array([
    [0.1, 0.05, 0.01],
    [0.05, 0.15, 0.04],
    [0.01, 0.04, 0.2]
])
```

```python
#Equilibrium portfolio based on market cap
w_mkt = np.array([0.5, 0.3, 0.2])

#Risk aversion parameter
delta = 2.25

# Equilibrium returns: π = δ * Σ * w
pi = delta * Sigma @ w_mkt
```

```python
#Tau: scaling factor
#Small values in the range 0.025 to 0.05 recommended
tau = 0.05

#Investor's belief
#E.g, Asset 1 will outperform Asset 2 by 3% (relative performance)
#P matrix encodes views: View on Asset1 - Asset2
P = np.array([
    [1, -1, 0]
])
#Q are expected returns
Q = np.array([0.03])  # Investor expects +3%

#View uncertainty: common to assume proportional to variance
Omega = np.diag([0.1**2])
```

```python
#BL posterior returns
M = np.linalg.inv(P @ (tau * Sigma) @ P.T + Omega)
adj = (tau * Sigma) @ P.T @ M @ (Q - P @ pi)
bl_returns = pi + adj

#Posterior covariance
M = np.linalg.inv(np.linalg.inv(tau * Sigma) + P.T @ np.linalg.inv(Omega) @ P)
bl_cov = Sigma + M

for idx,r in enumerate(pi):
	print(f"Equilibrium return Asset {idx+1}: {r:.3f}")

for idx,r in enumerate(bl_returns):
	print(f"Posterior return: Asset {idx+1}: {r:.3f}")
```
