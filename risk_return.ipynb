### Monthly Cumulative Returns for SPY ETF

```python
"""
Mean-Variance Optimisation
"""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import minimize
import yfinance as yf
```

```python
#DJIA constituents as of 2025-05-29
TICKERS = [
    "MMM",  # 3M
    "AXP",  # American Express
    "AMGN", # Amgen
    "AMZN", # Amazon
    "AAPL", # Apple
    "BA",   # Boeing
    "CAT",  # Caterpillar
    "CVX",  # Chevron
    "CSCO", # Cisco
    "KO",   # Coca-Cola
    "DIS",  # Disney
    "GS",   # Goldman Sachs
    "HD",   # Home Depot
    "HON",  # Honeywell
    "IBM",  # IBM
    "JNJ",  # Johnson & Johnson
    "JPM",  # JPMorgan Chase
    "MCD",  # McDonald's
    "MRK",  # Merck
    "MSFT", # Microsoft
    "NKE",  # Nike
    "NVDA", # Nvidia
    "PG",   # Procter & Gamble
    "CRM",  # Salesforce
    "SHW",  # Sherwin-Williams
    "TRV",  # Travelers
    "UNH",  # United Health Group
    "VZ",   # Verizon
    "V",    # Visa
    "WMT",  # Walmart
]
```

```python
def data_download(TICKERS: list, period="1y", interval="1d"):
    """
    Parameters
    ----------
    TICKERS: list, tickers to download
    period: str, default one year "1y"
    interval: str, default one day "1d"

    Returns
    -------
    df_ret: DataFrame, daily log returns of stocks
    """
    #The group_by="ticker" returns a column hierarchy per ticker
    data = yf.download(
        tickers=TICKERS,
        period=period,
        interval=interval,
        auto_adjust=True,
        progress=False,
        group_by="ticker",
    )

    #Extract adjusted close
    df = pd.DataFrame({t: data[t]["Close"] if "Close" in data[t] else data[t]["Adj Close"]
         for t in TICKERS if t in data.columns.get_level_values(0)})

    #Compute daily log returns
    df_ret = np.log(df).diff().dropna()

    #Save data
    df_ret.to_csv("downloaded_data.csv", index=False)
    return df_ret


def moments(df, TRAD_DAYS=252):
    """
    Parameters
    ----------
    df:         Dataframe, log returns of n assets
    TRAD_DAYS:  int, number of trading days, default 252
    Returns
    -------
    assets:     list, names of securities
    n:          int, number of assets
    mu:         float, expected returns
    Sigma:      float, covariance matrix
    TRAD_DAYS:  int, number of trading days (for later calculations)
    """
    #Assumes daily data
    mu_annual = df.mean() * TRAD_DAYS
    Sigma_annual = df.cov() * TRAD_DAYS
    assets = df.columns.tolist()
    n = len(assets)
    mu = mu_annual.loc[assets].values
    Sigma = Sigma_annual.loc[assets, assets].values
    return TRAD_DAYS, assets, n, mu, Sigma


def portfolio_stats(w: np.ndarray, rf=0.04237):
    """
    Parameters
    ----------
    w:                  np.ndarray, portfolio weights
    rf:                 float, risk-free rate, default current 10-year US yield
    Returns
    -------
    ret, vol, sharpe:   tuple, (return, volatility, Sharpe ratio)
    """
    ret = float(w @ mu)
    vol = float(np.sqrt(w @ Sigma @ w))
    sharpe = (ret - rf) / vol if vol > 0 else np.nan
    return ret, vol, sharpe


def var_obj(w):
    """
    Parameters
    ----------
    w:                  np.ndarray, portfolio weights
    Returns
    -------
    Objective function  value of function, portfolio variance
    """
    return w @ Sigma @ w
```

```python
#Implementation
try:
    df_ret = pd.read_csv("downloaded_data.csv")
    print("File found")
except:
    print("Start download")
    df_ret = data_download(TICKERS)

#Check data structure
print(df_ret.head())

TRAD_DAYS, assets, n, mu, Sigma = moments(df_ret)

#Boundaries and constraints
bounds = [(0.0, 1.0)] * n
cons = ({"type": "eq", "fun": lambda w: np.sum(w) - 1.0},)

#Minimum-variance portfolio
#Initial values
w0 = np.repeat(1.0 / n, n)

res_minvar = minimize(var_obj, w0, method="SLSQP", bounds=bounds, constraints=cons)
w_minvar = res_minvar.x
ret_minvar, vol_minvar, sharpe_minvar = portfolio_stats(w_minvar)

#Efficient frontier
#Minimum-variance portfolios for a range of expected returns
target_rets = np.linspace(vol_minvar, max(mu), 30)
frontier = []
for target in target_rets:
    cons_tr = (
        {"type": "eq", "fun": lambda w: np.sum(w) - 1.0},
        {"type": "eq", "fun": lambda w, t=target: w @ mu - t},
    )
    res = minimize(var_obj, w0, method="SLSQP", bounds=bounds, constraints=cons_tr)
    if res.success:
        w = res.x
        r, v, s = portfolio_stats(w)
        frontier.append({"target": target, "ret": r, "vol": v, "sharpe": s})

frontier_df = pd.DataFrame(frontier)
```

```python
#Visualisation
plt.plot(frontier_df['vol'], frontier_df['ret'])
plt.ylabel('Volatility')
plt.xlabel('Expected return')
plt.show()
```
